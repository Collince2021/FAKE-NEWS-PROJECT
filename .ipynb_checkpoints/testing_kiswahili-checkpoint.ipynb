```python
import pandas as pd
import openai
from tqdm import tqdm

# Setting the OpenAI API key
openai.api_key = "your_api_key"

# Loading the dataset
df = pd.read_csv("True Clean.csv")

# Enabling tqdm for pandas
tqdm.pandas()

# Function to translate text using OpenAI API
def translate_text(text):
    if pd.isna(text):  # Handle NaN values
        return text
    try:
        response = openai.ChatCompletion.create(
            model="gpt-4",
            messages=[{"role": "system", "content": f"Translate this to Kiswahili: {text}"}]
        )
        return response["choices"][0]["message"]["content"]
    except Exception as e:
        print(f"Error translating: {text} - {e}")
        return text

# Apply translation using progress_apply
for col in df.select_dtypes(include=["object"]).columns:
    df[col] = df[col].progress_apply(translate_text)

# Saving translated dataset
df.to_csv("True_translated.csv", index=False)

print("Translation completed! Saved as True_translated.csv.")
```

```python
import pandas as pd
import openai
from tqdm import tqdm

# Set OpenAI API key
client = openai.OpenAI(api_key="your_api_key")  # Replace with your actual API key

# Load dataset
df = pd.read_csv("True Clean.csv")

# Function to translate text to Kiswahili
def translate_text(text):
    if pd.isna(text) or text.strip() == "":
        return text  # Skip empty values

    try:
        response = client.chat.completions.create(
            model="gpt-4",
            messages=[{"role": "system", "content": f"Translate this to Kiswahili: {text}"}]
        )
        return response.choices[0].message.content
    except Exception as e:
        print(f"Error translating: {text} - {e}")
        return text  # Return original text if error occurs

# Apply translation to all text columns
tqdm.pandas(disable=False)  # Enable progress bar

for i, col in enumerate(df.select_dtypes(include=["object"]).columns):
    print(f"Translating column {i+1}/{len(df.columns)}: {col}")
    df[col] = df[col].progress_apply(translate_text)

# Save translated dataset
df.to_csv("True_translated.csv", index=False)
print("Translation completed. File saved as 'True_translated.csv'.")
```
